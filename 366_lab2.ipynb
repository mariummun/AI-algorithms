```python
year = int(input("Enter year: "))

if (year % 4 == 0 and year % 100 != 0) :
  print(f'{year} is a leap year')
elif(year % 100 == 0 and year % 400 == 0):
  print(f'{year} is a leap year')
else :
  print(f'{year} is not a leap year')
```

exercise : 4

```python
num = [1, 6, 3, 8, 10, 2, 57]

max_num = num[0]

for i in num :
  if max_num < i:
    max_num = i
print(f'Max is {max_num}')

min_num = num[0]

for i in num :
  if min_num > i:
    min_num = i
print(f'Min is {min_num}')

sum = 0
for i in num:
  sum += i
print(f'Avg is', sum/len(num))
```

```python
students = {
    "alan" : 90,
    "bob" : 84,
    "colin" : 92,
    "dylan" : 92,
    "elle" : 73
}

highest_marks = max(students.values())
print("Highest marks is", highest_marks)

for names, marks in students.items() :
  if marks == highest_marks :
    print(f'{highest_marks} was obtained by {names}')
```

exercise : 6  Given a tuple of temperatures in Celsius, convert it to a list, add 5 to each value, and convert the
result back to a tuple. Print the original and modified tuples.

```python
temp_og_tuple = (30, 32, 29, 28, 25, 18, 38)
print("Original Tuple :", temp_og_tuple)

temp_list = list(temp_og_tuple)
print("Original tuple converted to list :", temp_list)
new_list = []

for val in temp_list :
  val = val + 5
  new_list.append(val)

temp_new_tuple = tuple(new_list)
print("New Tuple :", temp_new_tuple)

```
