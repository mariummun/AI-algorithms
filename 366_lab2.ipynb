variable type

```python
name = "Alice"
age = 25
print("Updated Name:", name)
print("Updated Age:", age)

print(f'{name} is of ', type(name))
print(f'{age} is of ', type(age))
```

type conversion

```python
num_string = '12'
num_integer = 23

updated_num_string = int(num_string)

print(f'sum is = ', num_integer + updated_num_string)
```

arithmetic & logical operators

```python
a = 5
b = 2
print("Multipliication is ", a*b)
print("division is ", a/b)
print("sum is ", a+b)
print("subtraction is ", a-b)
print("exponent is ", a**b)
print("modulo is ", a%b)
print("floor division is ", a//b)

print('\n(a == b) -->', a == b)
print('(a != b) -->', a != b)
print('(a > b) -->', a > b)
print('(a < b) -->', a < b)
print('(a >= b) -->', a >= b)
print('(a <= b) -->', a <= b)
```

for-loops using list

```python
fruits = ["apple", "banana", "cherry"]
colors = ["red", "yellow", "black"]

for fruit in fruits :
  for color in colors:
    print(fruit, color)
```

log and sin/cos/tan

```python
import math

print(math.sin(math.radians(30)))
print(math.cos(math.radians(30)))
print(math.tan(math.radians(30)))

print(math.log2(64))
print(math.log(math.e))

print(math.pi)
```

list(insert & append)

```python
colors = ['Red', 'Black', 'Green']

print("sliced : ", colors[-2:])

colors.append("Lavender");
print(colors)

colors.insert(0, "Sky blue")
print(colors)

print('green' in colors)
```

list <-> tuple

```python
my_tuple = (1, 2, 3, 'a', 'b', 'c')

print(my_tuple)

for item in my_tuple:
  print(item)

my_list = list(my_tuple)
print(my_list)
```

sets : add & remove

```python
toys = {'car', 'plane', 'animal'}

toys.add("dinosaur")
print(toys)

toys.remove("car")
print(toys)

print('car' in toys)
```
